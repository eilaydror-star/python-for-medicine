## אופרטורים

### חשבון

| אופרטור | פעולה | דוגמה | תוצאה |
|---|---|---|---|
| `+ - *` | חיבור, חיסור, כפל | `3 * 4` | `12` |
| `/` | חילוק (תמיד מחזיר float) | `10 / 4` | `2.5` |
| `//` | חילוק שלם (בלי שארית) | `10 // 4` | `2` |
| `%` | שארית חלוקה | `10 % 4` | `2` |
| `**` | חזקה | `2 ** 3` | `8` |

דוגמה: המרת 135 דקות של ניתוח לשעות ודקות:

</div>

````python
minutes = 135
print(minutes // 60, "hours and", minutes % 60, "minutes")
````

<div dir="rtl">

**פלט צפוי:**

```text
2 hours and 15 minutes
```

### השוואה (התוצאה תמיד True או False)

| אופרטור | משמעות |

### לוגיים: and, or, not

</div>

````python
temp = 38.5
pulse = 110
print(temp > 38 and pulse > 100)   # both conditions true?
print(temp > 39 or pulse > 100)    # at least one true?
print(not temp > 38)               # the opposite
````

## מחרוזות

מחרוזת היא רצף של תווים. אפשר לחבר מחרוזות עם `+`, לקבל את האורך עם `len`, ולגשת לתו לפי מיקום (**האינדקס הראשון הוא 0**):

</div>

````python
drug = "Paracetamol"
print(len(drug))     # number of characters
print(drug[0])       # first character
print(drug[-1])      # last character
print(drug[0:4])     # characters 0,1,2,3 (stops BEFORE 4)
````

### מתודות של מחרוזת

מתודה היא פונקציה ששייכת לאובייקט מסוים. כותבים אותה עם נקודה: `object.method()`. כך עובדים עם הרבה דברים בפייתון, לא רק עם מחרוזות.

</div>

````python
diagnosis = "  Type 2 Diabetes  "
print(diagnosis.strip())                 # remove spaces at both ends
print(diagnosis.strip().upper())         # all capitals
print(diagnosis.strip().lower())         # all small letters
print(diagnosis.replace("2", "1"))       # replace text
print("120/80".split("/"))               # cut into pieces
print("diabetes" in diagnosis.lower())   # does it contain?
````

<div dir="rtl">

**פלט צפוי:**

```text

TYPE 2 DIABETES
type 2 diabetes
  Type 1 Diabetes
['120', '80']
True
```

### f-string: הדרך הנוחה להרכיב טקסט

כותבים `f` לפני המירכאות, ושמים משתנים בתוך סוגריים מסולסלים. `:.1f` פירושו "הצג ספרה אחת אחרי הנקודה".

</div>

````python
name = "Dana"
bmi = 25.308641975
print(f"{name}'s BMI is {bmi:.1f}")
````

## תנאים: if, elif, else

תנאי מאפשר לתוכנית לקבל החלטה. שלושה דברים חובה: **נקודתיים** בסוף שורת התנאי, ו**הזחה** (4 רווחים או Tab) של השורות שתלויות בתנאי. ההזחה היא מה שאומר לפייתון אילו שורות שייכות לתנאי.

</div>

````python
temp = 38.2

if temp >= 38:
    print("Fever")
    print("Consider antipyretics")
elif temp >= 37.5:
    print("Subfebrile")
else:
    print("Normal")

print("Done")   # not indented: runs anyway
````

<div dir="rtl">

**פלט צפוי:**

```text
Fever
Consider antipyretics
Done
```

פייתון בודק את התנאים מלמעלה למטה, ומריץ רק את **הראשון** שמתקיים. `elif` (קיצור של else if) אפשר לכתוב כמה פעמים שרוצים, ו-`else` הוא אופציונלי.

### דוגמה: סיווג BMI לפי ארגון הבריאות העולמי

</div>

````python
bmi = 27.4

if bmi < 18.5:
    category = "Underweight"
elif bmi < 25:
    category = "Normal"
elif bmi < 30:
    category = "Overweight"
else:
    category = "Obese"

print(f"BMI {bmi} -> {category}")
````

<div dir="rtl">

**פלט צפוי:**

```text
BMI 27.4 -> Overweight
```

שים לב שאין צורך לכתוב `bmi >= 18.5 and bmi < 25`. אם הגענו ל-elif השני, כבר ידוע שהתנאי הראשון לא התקיים.

### ✏️ תרגיל 2.2 (בינוני)

המשתמש מקליד לחץ דם בפורמט `"135/85"`. הפרד את הערך הסיסטולי והדיאסטולי, ההמר אותם למספרים, והדפס "Hypertension" אם הסיסטולי 140 ומעלה **או** הדיאסטולי 90 ומעלה. אחרת הדפס "Not hypertensive".

</div>

````python
# תרגיל 2.2: כתוב כאן את הפתרון שלך
````

````python
#@title 👀 הצג פתרון לתרגיל 2.2 { display-mode: "form" }
# double-click this cell to see the solution

bp = input("Blood pressure (e.g. 135/85): ")
parts = bp.split("/")
systolic = int(parts[0])
diastolic = int(parts[1])

if systolic >= 140 or diastolic >= 90:
    print("Hypertension")
else:
    print("Not hypertensive")
````

### ✏️ תרגיל 2.3 (בינוני)

המר טמפרטורה מפרנהייט לצלזיוס לפי `C = (F - 32) * 5 / 9`, והדפס אם יש חום (38 מעלות צלזיוס ומעלה). בדוק עם 100.4.

</div>

````python
# תרגיל 2.3: כתוב כאן את הפתרון שלך
````
