## כתיבה לקובץ

`open` פותח קובץ. הפרמטר השני אומר מה עושים איתו: `"w"` כתיבה (מוחק תוכן קודם), `"a"` הוספה לסוף, `"r"` קריאה (ברירת המחדל). הצורה `with` סוגרת את הקובץ אוטומטית בסוף הבלוק, ולכן תמיד משתמשים בה.

</div>

````python
data = """name,age,glucose
Dana,54,132
Avi,38,94
Noa,67,128
Tom,45,101
"""

with open("patients.csv", "w") as f:
    f.write(data)
````

## קריאה מקובץ

</div>

````python
with open("patients.csv") as f:
    text = f.read()      # the whole file as one string
print(text)
````

<div dir="rtl">

**פלט צפוי:**

```text
name,age,glucose
Dana,54,132
Avi,38,94
Noa,67,128
Tom,45,101
```

בדרך כלל עדיף לקרוא שורה אחרי שורה, עם לולאה:

</div>

````python
with open("patients.csv") as f:
    for line in f:
        print(line.strip().split(","))
````

<div dir="rtl">

**פלט צפוי:**

```text
['name', 'age', 'glucose']
['Dana', '54', '132']
['Avi', '38', '94']
['Noa', '67', '128']
['Tom', '45', '101']
```

`strip` מוריד את תו ירידת השורה שבסוף כל שורה, ו-`split` חותך לפי פסיק.

## קבצי csv עם המודול csv

csv (Comma-Separated Values) הוא טבלה בפורמט טקסט: כל שורה היא רשומה, והעמודות מופרדות בפסיק. אקסל יודע לשמור ולפתוח קבצים כאלה. המודול `csv` מטפל במקרים מסובכים בשבילך, למשל פסיק בתוך שם.

`import` טוען מודול: אוסף של כלים מוכנים שמגיע עם פייתון. נראה את זה הרבה מעכשיו.

</div>

````python
import csv

with open("patients.csv") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
        break      # just the first row
````

<div dir="rtl">

**פלט צפוי:**

```text
{'name': 'Dana', 'age': '54', 'glucose': '132'}
```

`DictReader` הופך כל שורה למילון, כשהמפתחות הם הכותרות מהשורה הראשונה. נוח מאוד.

> ⚠️ **הכול מגיע כטקסט**  
> שים לב ל-`'54'` במירכאות. כל ערך שנקרא מקובץ הוא מחרוזת, וצריך להמיר עם `int` או `float` לפני שמחשבים.

</div>

````python
high = []
with open("patients.csv") as f:
    for row in csv.DictReader(f):
        if int(row["glucose"]) >= 126:
            high.append(row["name"])
print(high)
````

<div dir="rtl">

**פלט צפוי:**

```text
['Dana', 'Noa']
```

### כתיבת csv

</div>

````python
with open("high_glucose.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "glucose"])   # header
    writer.writerow(["Dana", 132])
    writer.writerow(["Noa", 128])
````

<div dir="rtl">

`newline=""` מונע שורות ריקות מיותרות בווינדוס. תמיד כדאי להוסיף אותו בכתיבת csv.

## קבצי json

json שומר מבנים מקוננים: מילונים בתוך רשימות בתוך מילונים. הוא נראה כמעט בדיוק כמו מילון של פייתון. זה הפורמט הנפוץ ביותר לנתונים שמגיעים מאתרים ומשרתים.

</div>

````python
import json

patient = {"name": "Dana", "age": 54, "diabetic": True,
           "meds": ["metformin", "aspirin"]}

with open("patient.json", "w") as f:
    json.dump(patient, f, indent=2)    # indent makes it readable
````

<div dir="rtl">

כך נראה הקובץ שנוצר:

**פלט צפוי:**

```text
{
  "name": "Dana",
  "age": 54,
  "diabetic": true,
  "meds": [
    "metformin",
    "aspirin"
  ]
}
```

וקריאה בחזרה. הפעם המספרים נשארים מספרים:

</div>

````python
with open("patient.json") as f:
    loaded = json.load(f)

print(loaded["meds"][1])
print(type(loaded))
````

<div dir="rtl">

**פלט צפוי:**

```text
aspirin
<class 'dict'>
```

| פונקציה | מה היא עושה |
|---|---|
| `json.dump(obj, f)` | כותבת אובייקט לקובץ |
| `json.load(f)` | קוראת אובייקט מקובץ |
| `json.dumps(obj)` | הופכת אובייקט למחרוזת (ה-s בסוף = string) |
| `json.loads(text)` | הופכת מחרוזת לאובייקט |

## כריית נתונים מהרשת (scraping)

Scraping הוא שליפת מידע מדף אינטרנט באופן אוטומטי. משתמשים בשתי ספריות: `requests` מוריד את הדף, ו-`BeautifulSoup` מפרק את ה-HTML כדי שאפשר יהיה לחפש בו. שתיהן מותקנות מראש ב-Colab.

</div>

````python
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/Hypertension"
response = requests.get(url, headers={"User-Agent": "student-project"})
soup = BeautifulSoup(response.text, "html.parser")

print(soup.title.text)              # the page title
for h in soup.find_all("h2")[:5]:   # first 5 section headings
    print("-", h.text)
````

<div dir="rtl">

השורה הראשונה בפלט תהיה `Hypertension - Wikipedia`, ואחריה כותרות הפרקים של הערך, למשל Signs and symptoms ו-Causes. הרשימה המדויקת תלויה במבנה הדף באותו יום. `find_all("h2")` מחזיר את כל התגים מסוג `h2`. באותה דרך אפשר לחפש טבלאות (`"table"`), קישורים (`"a"`) ועוד.

### ✏️ תרגיל 5.2 (בינוני)

קרא את `patients.csv` בעזרת `DictReader` והדפס את מספר המטופלים ואת הגיל הממוצע.

</div>

````python
# תרגיל 5.2: כתוב כאן את הפתרון שלך
````

````python
#@title 👀 הצג פתרון לתרגיל 5.2 { display-mode: "form" }
# double-click this cell to see the solution

import csv
ages = []
with open("patients.csv") as f:
    for row in csv.DictReader(f):
        ages.append(int(row["age"]))
print(f"Patients: {len(ages)}, mean age: {sum(ages) / len(ages):.1f}")

# expected output:
# Patients: 4, mean age: 51.0
````

### ✏️ תרגיל 5.3 (אתגר)

קרא את `patients.csv`, הוסף לכל מטופל שדה `"diabetic"` (True אם הגלוקוז 126 ומעלה), ושמור את כל הרשימה לקובץ `patients.json`.

</div>

````python
# תרגיל 5.3: כתוב כאן את הפתרון שלך
````

````python
#@title 👀 הצג פתרון לתרגיל 5.3 { display-mode: "form" }
# double-click this cell to see the solution

import csv, json

patients = []
with open("patients.csv") as f:
    for row in csv.DictReader(f):
        row["age"] = int(row["age"])
        row["glucose"] = int(row["glucose"])
        row["diabetic"] = row["glucose"] >= 126
        patients.append(row)

with open("patients.json", "w") as f:
    json.dump(patients, f, indent=2)

# פתח את הקובץ מהתיקייה של Colab כדי לוודא שהוא נראה כמו שציפית.
````
