## רשימות (list)

רשימה מחזיקה כמה ערכים לפי סדר, בתוך סוגריים מרובעים. היא **ניתנת לשינוי**: אפשר להוסיף, להסיר ולעדכן.

</div>

````python
glucose = [95, 110, 87, 142, 101]   # fasting glucose, mg/dL

print(glucose[0])      # first
print(glucose[-1])     # last
print(glucose[1:3])    # slice: index 1 and 2
print(len(glucose))    # how many
print(min(glucose), max(glucose), sum(glucose))
````

<div dir="rtl">

**פלט צפוי:**

```text
95
101
[110, 87]
5
87 142 535
```

### שינוי רשימה

</div>

````python
glucose.append(99)        # add at the end
glucose[0] = 96           # replace a value
glucose.remove(142)       # remove by value
glucose.sort()            # sort in place
print(glucose)

average = sum(glucose) / len(glucose)
print(f"Average: {average:.1f}")
````

<div dir="rtl">

**פלט צפוי:**

```text
[87, 96, 99, 101, 110]
Average: 98.6
```

רשימה יכולה להכיל כל סוג נתון, גם סוגים מעורבים, ואפילו רשימות אחרות.

## Tuple

כמו רשימה, אבל עם סוגריים עגולים, ו**אי אפשר לשנות** אותו אחרי שנוצר. מתאים לנתונים שאמורים להישאר קבועים, למשל קואורדינטות או זוג ערכים של לחץ דם.

</div>

````python
bp = (120, 80)
systolic, diastolic = bp    # "unpacking" into two variables
print(systolic)
# bp[0] = 130  ->  TypeError: 'tuple' object does not support item assignment
````

<div dir="rtl">

**פלט צפוי:**

```text
120
```

## קבוצות (set)

אוסף **בלי כפילויות ובלי סדר**, בסוגריים מסולסלים. שימושי כדי להוריד כפילויות ולמצוא מה משותף בין שני אוספים.

</div>

````python
ward_a = {"aspirin", "insulin", "metformin", "aspirin"}
ward_b = {"insulin", "warfarin"}

print(ward_a)               # duplicate "aspirin" removed
print(ward_a & ward_b)      # in both (intersection)
print(ward_a | ward_b)      # in either (union)
print(ward_a - ward_b)      # only in A
````

<div dir="rtl">

**פלט צפוי:**

```text
{'metformin', 'insulin', 'aspirin'}
{'insulin'}
{'metformin', 'insulin', 'warfarin', 'aspirin'}
{'metformin', 'aspirin'}
```

הסדר בפלט שלך עשוי להיות שונה. זה תקין, לקבוצה אין סדר.

## מילונים (dict)

מילון שומר זוגות של **מפתח: ערך**. במקום לגשת לפי מיקום (0, 1, 2), ניגשים לפי שם. זה המבנה הטבעי לתיק מטופל.

</div>

````python
patient = {
    "name": "Dana Cohen",
    "age": 54,
    "diabetic": True,
    "meds": ["metformin", "aspirin"],
}

print(patient["name"])
print(patient["meds"][0])

patient["age"] = 55             # update
patient["allergy"] = "penicillin"   # add a new key
print(patient.keys())
print("allergy" in patient)
print(patient.get("weight", "unknown"))   # safe access with a default
````

<div dir="rtl">

**פלט צפוי:**

```text
Dana Cohen
metformin
dict_keys(['name', 'age', 'diabetic', 'meds', 'allergy'])
True
unknown
```

> ⚠️ **טעות נפוצה**  
> גישה למפתח שלא קיים, כמו `patient["weight"]`, תיתן `KeyError`. אם אתה לא בטוח שהמפתח קיים, השתמש ב-`.get`.

### מתי להשתמש במה

| מבנה | תחביר | סדר | ניתן לשינוי | כפילויות | מתאים ל... |
|---|---|---|---|---|---|
| list | `[1, 2]` | כן | כן | מותר | סדרת מדידות |
| tuple | `(1, 2)` | כן | לא | מותר | ערכים קבועים |
| set | `{1, 2}` | לא | כן | לא | ערכים ייחודיים |
| dict | `{"a": 1}` | כן | כן | מפתחות ייחודיים | רשומה עם שדות |

## לולאת for

לולאת `for` חוזרת על אותם צעדים עבור כל פריט באוסף. כמו ב-if: נקודתיים בסוף, והזחה לגוף הלולאה.

</div>

````python
temps = [36.8, 38.4, 37.1, 39.0]

for t in temps:
    if t >= 38:
        print(t, "fever")
    else:
        print(t, "ok")
````

<div dir="rtl">

**פלט צפוי:**

```text
36.8 ok
38.4 fever
37.1 ok
39.0 fever
```

### range: לולאה לפי מספרים

`range(5)` נותן 0, 1, 2, 3, 4 (לא כולל 5). `range(1, 6)` נותן 1 עד 5.

</div>

````python
for day in range(1, 4):
    print(f"Day {day}: take 1 pill")
````

<div dir="rtl">

**פלט צפוי:**

```text
Day 1: take 1 pill
Day 2: take 1 pill
Day 3: take 1 pill
```

### enumerate: לקבל גם את המיקום

### תבנית חשובה: צבירה

מתחילים עם משתנה ריק (0 או רשימה ריקה), ובכל סיבוב של הלולאה מוסיפים אליו. זו התבנית הכי נפוצה בתרגילים.

</div>

````python
temps = [36.8, 38.4, 37.1, 39.0]
fever_count = 0
fevers = []

for t in temps:
    if t >= 38:
        fever_count += 1      # same as fever_count = fever_count + 1
        fevers.append(t)

print(fever_count, fevers)
````

<div dir="rtl">

**פלט צפוי:**

```text
2 [38.4, 39.0]
```

## לולאת while

`while` חוזרת כל עוד תנאי מתקיים. משתמשים בה כשלא יודעים מראש כמה פעמים צריך לחזור.

</div>

````python
# a drug's concentration halves every 6 hours; when is it below 10?
conc = 100
hours = 0
while conc >= 10:
    conc = conc / 2
    hours += 6
print(f"After {hours} hours: {conc}")
````

<div dir="rtl">

**פלט צפוי:**

```text
After 24 hours: 6.25
```

> ⚠️ **לולאה אינסופית**  
> אם התנאי אף פעם לא נהיה False, הלולאה לא תיעצר. ב-Colab לוחצים על כפתור העצירה ליד התא. ודא שמשהו בתוך הלולאה משנה את התנאי.

`break` יוצא מהלולאה מיד. `continue` מדלג לסיבוב הבא.

## List comprehension (קיצור שימושי)

דרך לבנות רשימה חדשה בשורה אחת. אין חובה להשתמש בזה, אבל תראה את זה הרבה בקוד של אחרים.

</div>

````python
temps_f = [98.6, 101.3, 99.5]
temps_c = [round((f - 32) * 5 / 9, 1) for f in temps_f]
print(temps_c)

high = [t for t in temps_c if t >= 38]
print(high)
````

<div dir="rtl">

**פלט צפוי:**

```text
[37.0, 38.5, 37.5]
[38.5]
```

### ✏️ תרגיל 3.2 (בינוני)

על אותה רשימה: צור שתי רשימות חדשות, `tachy` לערכים מעל 100 ו-`brady` לערכים מתחת ל-60. השתמש בלולאת for.

</div>

````python
# תרגיל 3.2: כתוב כאן את הפתרון שלך
````

````python
#@title 👀 הצג פתרון לתרגיל 3.2 { display-mode: "form" }
# double-click this cell to see the solution

hr = [72, 88, 110, 65, 54, 101]
tachy = []
brady = []
for h in hr:
    if h > 100:
        tachy.append(h)
    elif h < 60:
        brady.append(h)
print("Tachycardia:", tachy)
print("Bradycardia:", brady)

# expected output:
# Tachycardia: [110, 101]
# Bradycardia: [54]
````

### ✏️ תרגיל 3.3 (בינוני)

נתונה רשימה של אבחנות: `["flu", "asthma", "flu", "covid", "flu", "asthma"]`. בנה מילון שסופר כמה פעמים הופיעה כל אבחנה.

</div>

````python
# תרגיל 3.3: כתוב כאן את הפתרון שלך
````

````python
#@title 👀 הצג פתרון לתרגיל 3.3 { display-mode: "form" }
# double-click this cell to see the solution

diagnoses = ["flu", "asthma", "flu", "covid", "flu", "asthma"]
counts = {}
for d in diagnoses:
    if d in counts:
        counts[d] += 1
    else:
        counts[d] = 1
print(counts)

# expected output:
# {'flu': 3, 'asthma': 2, 'covid': 1}
# זו תבנית שחוזרת הרבה. בהמשך נראה ש-pandas עושה את זה בפקודה אחת.
````

### ✏️ תרגיל 3.4 (אתגר)

נתונה רשימה של מטופלים, כל אחד מילון:

</div>

````python
patients = [
    {"name": "Dana", "age": 54, "glucose": 132},
    {"name": "Avi", "age": 38, "glucose": 94},
    {"name": "Noa", "age": 67, "glucose": 128},
    {"name": "Tom", "age": 45, "glucose": 101},
]
````

<div dir="rtl">

הדפס את שמות המטופלים שהגלוקוז שלהם 126 ומעלה (סף לסוכרת בצום), ואת הגיל הממוצע שלהם.

</div>

````python
# תרגיל 3.4: כתוב כאן את הפתרון שלך
````

````python
#@title 👀 הצג פתרון לתרגיל 3.4 { display-mode: "form" }
# double-click this cell to see the solution

ages = []
for p in patients:
    if p["glucose"] >= 126:
        print(p["name"])
        ages.append(p["age"])
print("Average age:", sum(ages) / len(ages))

# expected output:
# Dana
# Noa
# Average age: 60.5
````
