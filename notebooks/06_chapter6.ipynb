## matplotlib: הבסיס

matplotlib היא ספריית הגרפים הוותיקה של פייתון, ורוב הספריות האחרות בנויות עליה. מייבאים את החלק שלה שנקרא `pyplot`, ונהוג לקצר אותו ל-`plt`:

</div>

```python
import matplotlib.pyplot as plt
```

<div dir="rtl">

כל גרף בנוי באותו מבנה: יוצרים, מוסיפים כותרות, ומציגים. ב-Colab הגרף מופיע מתחת לתא גם בלי `plt.show()`, אבל מקובל לכתוב את זה.

## גרף קו: שינוי לאורך זמן

</div>

```python
hours = [0, 2, 4, 6, 8, 10, 12]
heart_rate = [72, 75, 90, 110, 98, 85, 78]

plt.plot(hours, heart_rate, marker="o", color="crimson")
plt.title("Heart rate after surgery")
plt.xlabel("Hours")
plt.ylabel("BPM")
plt.grid(True)
plt.show()
```

<div dir="rtl">

הרשימה הראשונה היא ציר ה-x והשנייה ציר ה-y. שתיהן חייבות להיות באותו אורך.

## גרף פיזור: קשר בין שני משתנים

</div>

```python
age = [25, 32, 41, 47, 53, 58, 62, 68, 74, 80]
systolic = [115, 118, 124, 130, 128, 138, 142, 145, 150, 158]

plt.scatter(age, systolic)
plt.title("Systolic BP by age")
plt.xlabel("Age")
plt.ylabel("Systolic BP (mmHg)")
plt.show()
```

## גרף עמודות: השוואה בין קטגוריות

</div>

```python
diagnoses = ["Flu", "Asthma", "COVID", "Diabetes"]
counts = [42, 18, 25, 31]

plt.bar(diagnoses, counts, color="teal")
plt.title("Patients per diagnosis")
plt.ylabel("Patients")
plt.show()
```

## היסטוגרמה: איך הערכים מתפלגים

היסטוגרמה מחלקת את טווח הערכים ל"תאים" (bins) וסופרת כמה ערכים נופלים בכל תא. כך רואים את צורת ההתפלגות. כאן נייצר 500 ערכי גלוקוז אקראיים (על מספרים אקראיים נלמד בפרק 8):

</div>

```python
import random
random.seed(0)
glucose = [random.gauss(100, 15) for _ in range(500)]

plt.hist(glucose, bins=25, edgecolor="black")
plt.axvline(126, color="red", linestyle="--", label="Diabetes threshold")
plt.legend()
plt.xlabel("Fasting glucose (mg/dL)")
plt.ylabel("Patients")
plt.show()
```

<div dir="rtl">

`axvline` מצייר קו אנכי, ו-`label` יחד עם `legend()` מוסיפים מקרא.

## כמה גרפים זה לצד זה

</div>

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(9, 3.5))   # 1 row, 2 columns

ax1.plot(hours, heart_rate, marker="o")
ax1.set_title("Heart rate")

ax2.bar(diagnoses, counts)
ax2.set_title("Diagnoses")

plt.tight_layout()   # prevents overlapping labels
plt.show()
```

<div dir="rtl">

שים לב: על `ax` כותבים `set_title` ולא `title`. `figsize` קובע רוחב וגובה באינצ'ים.

## seaborn: גרפים סטטיסטיים בשורה אחת

seaborn בנויה על matplotlib ומקצרת גרפים סטטיסטיים. היא עובדת טוב עם טבלאות (נלמד pandas בפרק 10), אבל מקבלת גם מילון רגיל.

</div>

```python
import seaborn as sns

data = {
    "group": ["Placebo"] * 6 + ["Drug"] * 6,
    "systolic": [148, 152, 145, 160, 150, 155, 132, 138, 128, 141, 135, 130],
}
sns.boxplot(x="group", y="systolic", data=data)
plt.title("Systolic BP after 8 weeks")
plt.show()
```

<div dir="rtl">

תרשים קופסה (boxplot) מראה את החציון (הקו באמצע), את הרבעונים (גבולות הקופסה) ואת הטווח. נפוץ מאוד במאמרים רפואיים להשוואה בין קבוצות. גרפים שימושיים נוספים: `sns.histplot`, `sns.scatterplot`, `sns.heatmap`.

### ✏️ תרגיל 6.3 (אתגר)

קרא את `patients.csv` מפרק 5 וצייר גרף עמודות של הגלוקוז לפי שם מטופל. צבע באדום עמודות של 126 ומעלה ובירוק את השאר (רמז: `color` יכול לקבל רשימת צבעים).

הרץ קודם את התא הזה. הוא מכין את הנתונים שהתרגיל צריך.

</div>

```python
import matplotlib.pyplot as plt

# creates patients.csv (from chapter 5)
data = """name,age,glucose
Dana,54,132
Avi,38,94
Noa,67,128
Tom,45,101
"""
with open("patients.csv", "w") as f:
    f.write(data)
```

```python
# תרגיל 6.3: כתוב כאן את הפתרון שלך
```

```python
#@title 👀 הצג פתרון לתרגיל 6.3 { display-mode: "form" }
# double-click this cell to see the solution

import csv
names, values = [], []
with open("patients.csv") as f:
    for row in csv.DictReader(f):
        names.append(row["name"])
        values.append(int(row["glucose"]))

colors = ["red" if v >= 126 else "green" for v in values]
plt.bar(names, values, color=colors)
plt.ylabel("Glucose (mg/dL)")
plt.title("Fasting glucose by patient")
plt.show()
```
