## למה צריך פונקציות

כבר השתמשת בפונקציות מוכנות: `print`, `len`, `sum`. פונקציה היא קטע קוד עם שם, שאפשר להפעיל שוב ושוב. במקום להעתיק את חישוב ה-BMI עשר פעמים, כותבים אותו פעם אחת.

## הגדרת פונקציה

</div>

````python
def calc_bmi(weight, height):
    bmi = weight / height ** 2
    return bmi

result = calc_bmi(82, 1.80)
print(round(result, 1))
print(round(calc_bmi(60, 1.65), 1))
````

<div dir="rtl">

**פלט צפוי:**

```text
25.3
22.0
```

מה יש כאן:

`def` פותח הגדרה של פונקציה. אחריו השם, ובסוגריים **הפרמטרים**: השמות שיקבלו את הערכים שנעביר. בסוף נקודתיים, וגוף הפונקציה מוזח.  

`return` מחזיר תוצאה למי שקרא לפונקציה, והפונקציה מסתיימת באותו רגע.  

  הגדרת הפונקציה לא מריצה אותה. היא רצה רק כשקוראים לה: `calc_bmi(82, 1.80)`.

> ⚠️ **print מול return**  
> זה הבלבול הנפוץ ביותר. `print` רק מציג על המסך. `return` מחזיר ערך שאפשר לשמור במשתנה ולהמשיך לחשב איתו. פונקציה בלי return מחזירה `None` (כלום).

</div>

````python
def bmi_print(w, h):
    print(w / h ** 2)

x = bmi_print(82, 1.80)   # prints, but returns nothing
print(x)
````

**פלט צפוי:**

```text

None
```

## פונקציה שמחזירה טקסט

</div>

````python
def bmi_category(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    return "Obese"

print(bmi_category(calc_bmi(82, 1.80)))   # a function inside a function call
````

<div dir="rtl">

**פלט צפוי:**

```text
Overweight
```

שים לב: אין צורך ב-else בסוף. אם הגענו לשורה האחרונה, אף return קודם לא הופעל.

## ערכי ברירת מחדל ושמות פרמטרים

</div>

````python
def dose(weight_kg, mg_per_kg=15):
    """Return a weight-based dose in mg."""
    return weight_kg * mg_per_kg

print(dose(20))                  # uses the default 15
print(dose(20, 10))              # overrides it
print(dose(mg_per_kg=10, weight_kg=20))   # by name, any order
````

<div dir="rtl">

**פלט צפוי:**

```text
300
200
200
```

הטקסט בשלוש מירכאות מתחת לשורת ה-def נקרא **docstring**. הוא מתעד מה הפונקציה עושה, ו-Colab מציג אותו כשמרחפים מעל שם הפונקציה.

## החזרת כמה ערכים

</div>

````python
def stats(values):
    mean = sum(values) / len(values)
    return min(values), max(values), mean    # returns a tuple

lo, hi, avg = stats([95, 110, 87, 142])
print(lo, hi, avg)
````

```text

```

## משתנים מקומיים וגלובליים

משתנה שנוצר **בתוך** פונקציה קיים רק בתוכה (מקומי). מבחוץ הוא לא מוכר. משתנה שנוצר מחוץ לכל פונקציה הוא גלובלי, ואפשר לקרוא אותו גם מתוך פונקציה.

</div>

````python
threshold = 126            # global

def is_diabetic(glucose):
    result = glucose >= threshold    # reading a global is fine
    return result

print(is_diabetic(130))
print(result)              # result is local -> error
````

**פלט צפוי:**


> 💡 **כלל אצבע**  
> פונקציה טובה מקבלת את כל מה שהיא צריכה דרך הפרמטרים ומחזירה תוצאה עם return. היא לא תלויה במשתנים גלובליים ולא משנה אותם. כך קל לבדוק אותה ולהשתמש בה שוב.

## פונקציה שעוברת על רשימה

</div>

````python
def count_above(values, limit):
    count = 0
    for v in values:
        if v > limit:
            count += 1
    return count

hr = [72, 88, 110, 65, 54, 101]
print(count_above(hr, 100))
````

**פלט צפוי:**

```text
2
```

### ✏️ תרגיל 4.2 (בינוני)

כתוב פונקציה `bp_category(sys, dia)` שמחזירה "Normal" אם סיסטולי מתחת ל-120 וגם דיאסטולי מתחת ל-80, "Hypertension" אם סיסטולי 140 ומעלה או דיאסטולי 90 ומעלה, ו-"Elevated" בכל מקרה אחר.

````python
#@title 👀 הצג פתרון לתרגיל 4.2 { display-mode: "form" }
# double-click this cell to see the solution

def bp_category(sys, dia):
    if sys >= 140 or dia >= 90:
        return "Hypertension"
    if sys < 120 and dia < 80:
        return "Normal"
    return "Elevated"

print(bp_category(115, 75))
print(bp_category(130, 85))
print(bp_category(150, 70))

# expected output:
# Normal
# Elevated
# Hypertension
# הסדר חשוב: בודקים קודם את המקרה החמור.
````

### ✏️ תרגיל 4.3 (בינוני)

כתוב פונקציה `egfr_stage(egfr)` שמחזירה את שלב מחלת הכליות לפי eGFR: 90 ומעלה "G1", 60 עד 89 "G2", 45 עד 59 "G3a", 30 עד 44 "G3b", 15 עד 29 "G4", מתחת ל-15 "G5". הפעל אותה בלולאה על `[95, 72, 50, 33, 20, 9]`.

</div>

````python
# תרגיל 4.3: כתוב כאן את הפתרון שלך
````

````python
#@title 👀 הצג פתרון לתרגיל 4.3 { display-mode: "form" }
# double-click this cell to see the solution

def egfr_stage(egfr):
    if egfr >= 90:
        return "G1"
    elif egfr >= 60:
        return "G2"
    elif egfr >= 45:
        return "G3a"
    elif egfr >= 30:
        return "G3b"
    elif egfr >= 15:
        return "G4"
    return "G5"

for v in [95, 72, 50, 33, 20, 9]:
    print(v, egfr_stage(v))

# expected output:
# 95 G1
# 72 G2
# 50 G3a
# 33 G3b
# 20 G4
# 9 G5
````

### ✏️ תרגיל 4.4 (אתגר)

כתוב פונקציה `summarize(patients)` שמקבלת רשימת מילונים כמו בתרגיל 3.4, ומחזירה מילון עם מספר המטופלים, הגיל הממוצע, ורשימת שמות של מי שהגלוקוז שלו 126 ומעלה.

הרץ קודם את התא הזה. הוא מכין את הנתונים שהתרגיל צריך.

</div>

````python
patients = [
    {"name": "Dana", "age": 54, "glucose": 132},
    {"name": "Avi", "age": 38, "glucose": 94},
    {"name": "Noa", "age": 67, "glucose": 128},
    {"name": "Tom", "age": 45, "glucose": 101},
]
````

````python
# תרגיל 4.4: כתוב כאן את הפתרון שלך
````

````python
#@title 👀 הצג פתרון לתרגיל 4.4 { display-mode: "form" }
# double-click this cell to see the solution

def summarize(patients):
    ages = []
    high = []
    for p in patients:
        ages.append(p["age"])
        if p["glucose"] >= 126:
            high.append(p["name"])
    return {
        "count": len(patients),
        "mean_age": sum(ages) / len(ages),
        "high_glucose": high,
    }

print(summarize(patients))

# expected output:
# {'count': 4, 'mean_age': 51.0, 'high_glucose': ['Dana', 'Noa']}
````
